```python
!pip install tensorflow==2.15 torch torchvision torchaudio matplotlib

# --- Imports ---
import tensorflow as tf
import torch
import torch.nn as nn
import torch.optim as optim
import matplotlib.pyplot as plt

# ==============================
# THEORETICAL PART (Q&A)
# ==============================

theory_answers = {
    "1. What is TensorFlow 2.0, and how is it different from TensorFlow 1.x":
    "TensorFlow 2.0 is a major upgrade over 1.x, focusing on simplicity and ease of use. "
    "It integrates Keras as the high-level API, uses eager execution by default, "
    "and removes redundant APIs present in 1.x.",

    "2. How do you install TensorFlow 2.0":
    "You can install TensorFlow 2.0 in Colab using: pip install tensorflow==2.15",

    "3. What is the primary function of tf.function in TensorFlow 2.0":
    "It converts Python functions into high-performance TensorFlow graphs, improving execution speed.",

    "4. What is the purpose of the Model class in TensorFlow 2.0":
    "It is used to build deep learning models by defining layers, training, and evaluation steps.",

    "5. What is the importance of Tensor Space in TensorFlow":
    "Tensor Space represents the multi-dimensional data (scalars, vectors, matrices, tensors) "
    "on which computations are performed.",

    "6. How can TensorBoard be integrated with TensorFlow 2.0":
    "By using tf.summary and callbacks during model training, we can visualize metrics and graphs.",

    "7. What is TensorFlow Playground":
    "It’s an interactive web-based tool to visualize and experiment with neural networks.",

    "8. What is Netron, and how is it useful":
    "Netron is a visualizer for neural network, deep learning, and machine learning models.",

    "9. Difference between TensorFlow and PyTorch":
    "TensorFlow is more production-oriented with deployment support, while PyTorch is more developer-friendly "
    "with dynamic computation graphs.",

    "10. What is the purpose of the torch.optim module in PyTorch":
    "It provides optimization algorithms like SGD, Adam, RMSProp for training models.",

    "11. Common activation functions":
    "ReLU, Sigmoid, Tanh, Softmax are commonly used.",

    "12. Difference between torch.nn.Module and torch.nn.Sequential":
    "nn.Module allows custom model definitions; nn.Sequential stacks layers sequentially.",

    "13. How does Keras fit into TensorFlow 2.0":
    "Keras is the official high-level API integrated into TensorFlow 2.0."
}

for q, a in theory_answers.items():
    print(q)
    print("Answer:", a, "\n")

# ==============================
# PRACTICAL PART
# ==============================

# --- Verify TensorFlow Installation ---
print("TensorFlow Version:", tf.__version__)

# --- Simple TensorFlow Function ---
@tf.function
def add_fn(x, y):
    return x + y

print("Addition using tf.function:", add_fn(tf.constant(3), tf.constant(5)).numpy())

# --- Simple Neural Network in TensorFlow ---
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense

model_tf = Sequential([
    Dense(8, activation='relu', input_shape=(4,)),
    Dense(1, activation='sigmoid')
])

model_tf.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
print("TensorFlow model summary:")
model_tf.summary()

# --- Simple Visualization Example ---
epochs = list(range(1, 11))
loss = [1/(e+1) for e in epochs]
plt.plot(epochs, loss, marker='o')
plt.title("Training Loss Example")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.show()

# --- Verify PyTorch Installation ---
print("PyTorch Version:", torch.__version__)

# --- Simple PyTorch Neural Network ---
class SimpleNN(nn.Module):
    def __init__(self):
        super(SimpleNN, self).__init__()
        self.fc1 = nn.Linear(4, 8)
        self.relu = nn.ReLU()
        self.fc2 = nn.Linear(8, 1)
        self.sigmoid = nn.Sigmoid()

    def forward(self, x):
        out = self.fc1(x)
        out = self.relu(out)
        out = self.fc2(out)
        out = self.sigmoid(out)
        return out

model_torch = SimpleNN()
print("PyTorch Model:\n", model_torch)

# --- Loss & Optimizer in PyTorch ---
criterion = nn.BCELoss()
optimizer = optim.Adam(model_torch.parameters(), lr=0.01)

print("Loss function:", criterion)
print("Optimizer:", optimizer)
```
