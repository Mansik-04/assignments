What is K-Nearest Neighbors (KNN) and how does it work
KNN is a lazy, instance-based supervised algorithm. For a new point it finds the K closest training points (by a distance metric) and predicts by majority vote (classification) or average (regression).

Difference between KNN Classification and KNN Regression
Classification → majority vote of neighbors’ labels.
Regression → average (or weighted average) of neighbors’ target values.

Role of the distance metric in KNN
Distance defines “closeness”. Common metrics: Euclidean, Manhattan. Metric choice influences which neighbors are selected and therefore predictions.

Curse of Dimensionality in KNN
As dimensionality grows, distances become less meaningful (points spread out), degrading KNN performance. Dimensionality reduction or feature selection helps.

How to choose best K
Use cross-validation or plot validation error vs K. Small K → low bias/high variance; large K → high bias/low variance.

KD-Tree and Ball-Tree
Both are spatial data structures to accelerate neighbor queries. KD-Tree partitions along axis-aligned hyperplanes; Ball-Tree partitions with hyperspheres.

When KD-Tree vs Ball-Tree
KD-Tree: good for low to moderate dimensions (d ≲ ~20). Ball-Tree: often better for higher dimensions or non-axis-aligned clusters.

Disadvantages of KNN
Slow at prediction (unless indexing), sensitive to irrelevant features & scaling, poor in high-D, storage heavy.

How feature scaling affects KNN
Essential — distance-based; features must be scaled (StandardScaler/MinMax) so no single feature dominates.

What is PCA (Principal Component Analysis)
Unsupervised linear dimensionality reduction that finds orthogonal axes (principal components) capturing maximum variance.

How PCA works
Center data → compute covariance matrix → eigen-decomposition → project onto top eigenvectors (principal components).

Geometric intuition behind PCA
Find directions (lines/planes) that best preserve variance (closest low-dimensional subspace to data in least-squares sense).

Feature Selection vs Feature Extraction
Selection: pick subset of original features. Extraction: create new features (PCA) by transforming originals.

Eigenvalues and Eigenvectors in PCA
Eigenvectors = principal directions; eigenvalues = variance explained along those directions.

Decide number of components
Use explained variance ratio or cumulative explained variance (e.g., keep components explaining ≥90–95% variance).

Can PCA be used for classification
PCA is unsupervised; it can be used as preprocessing for classification (reduce noise/Dim) but doesn’t use labels.

Limitations of PCA
Linear only, sensitive to scaling, may mix features making interpretability harder, not optimal for classification objective.

How KNN and PCA complement each other
PCA reduces dimensionality/noise → KNN performs better (faster queries, less curse-of-dim).

How KNN handles missing values
Typical approach: impute missing values (mean/median/KNN imputation) before applying KNN.

Differences between PCA and LDA
PCA: unsupervised, maximizes variance.
LDA: supervised, maximizes class separability (between-class / within-class variance).

```python
# KNN & PCA — All-in-one notebook
# Run in Python 3.8+, needs scikit-learn, numpy, pandas, matplotlib, seaborn

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_iris, load_wine, make_regression, load_breast_cancer
from sklearn.model_selection import train_test_split, GridSearchCV, cross_val_score
from sklearn.neighbors import KNeighborsClassifier, KNeighborsRegressor
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.decomposition import PCA
from sklearn.metrics import accuracy_score, mean_squared_error, classification_report, confusion_matrix, roc_auc_score, precision_recall_fscore_support

from sklearn.impute import KNNImputer

RANDOM_STATE = 42
sns.set(style="whitegrid")

# ----------------------
# Helper plotting utils
# ----------------------
def plot_decision_boundary(clf, X, y, title="Decision boundary (2 features)"):
    # X should be 2D
    x_min, x_max = X[:,0].min()-1, X[:,0].max()+1
    y_min, y_max = X[:,1].min()-1, X[:,1].max()+1
    xx, yy = np.meshgrid(np.linspace(x_min,x_max,300), np.linspace(y_min,y_max,300))
    Z = clf.predict(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)
    plt.figure(figsize=(6,5))
    plt.contourf(xx, yy, Z, alpha=0.3)
    plt.scatter(X[:,0], X[:,1], c=y, edgecolor='k', s=40)
    plt.title(title)
    plt.show()

# =========================
# THEORETICAL -> PRACTICAL
# =========================

# 1) KNN Classifier on Iris dataset and print accuracy
iris = load_iris()
X, y = iris.data, iris.target
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=RANDOM_STATE, stratify=y)

knn = KNeighborsClassifier(n_neighbors=5, metric='minkowski')   # Euclidean by default (p=2)
knn.fit(X_train, y_train)
y_pred = knn.predict(X_test)
print("1) KNN (k=5) Iris Accuracy:", accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred, target_names=iris.target_names))

# 2) KNN Regressor on synthetic data (MSE)
Xr, yr = make_regression(n_samples=500, n_features=5, noise=10.0, random_state=RANDOM_STATE)
Xr_train, Xr_test, yr_train, yr_test = train_test_split(Xr, yr, test_size=0.2, random_state=RANDOM_STATE)
knn_reg = KNeighborsRegressor(n_neighbors=5)
knn_reg.fit(Xr_train, yr_train)
yr_pred = knn_reg.predict(Xr_test)
print("2) KNN Regressor MSE:", mean_squared_error(yr_test, yr_pred))

# 3) KNN with Euclidean vs Manhattan metrics (compare accuracy)
knn_euc = KNeighborsClassifier(n_neighbors=5, metric='euclidean')
knn_man = KNeighborsClassifier(n_neighbors=5, metric='manhattan')
knn_euc.fit(X_train, y_train)
knn_man.fit(X_train, y_train)
print("3) Euclidean accuracy:", accuracy_score(y_test, knn_euc.predict(X_test)))
print("   Manhattan accuracy:", accuracy_score(y_test, knn_man.predict(X_test)))

# 4) KNN decision boundaries for different K (use first two features of Iris)
X2 = iris.data[:, :2]; y2 = iris.target
X2_train, X2_test, y2_train, y2_test = train_test_split(X2, y2, test_size=0.2, random_state=RANDOM_STATE, stratify=y2)
for k in [1,3,7,15]:
    clf = KNeighborsClassifier(n_neighbors=k)
    clf.fit(X2_train, y2_train)
    print(f"4) K={k} accuracy (2-feat):", accuracy_score(y2_test, clf.predict(X2_test)))
    plot_decision_boundary(clf, X2_train, y2_train, title=f"KNN decision boundary (k={k})")

# 5) Feature scaling effect before training KNN: compare scaled vs unscaled
# Create a toy dataset where features have different scales
X_toy = np.hstack([np.random.normal(size=(200,1))*0.01, np.random.normal(size=(200,1))])  # one small-scale, one large-scale
y_toy = np.array([0]*100 + [1]*100)
X_toy[:100, :] += np.array([0.0, -1.5])
X_toy[100:, :] += np.array([0.0, 1.5])
X_train_t, X_test_t, y_train_t, y_test_t = train_test_split(X_toy, y_toy, test_size=0.2, random_state=RANDOM_STATE, stratify=y_toy)

knn_unscaled = KNeighborsClassifier(n_neighbors=5)
knn_unscaled.fit(X_train_t, y_train_t)
acc_unscaled = accuracy_score(y_test_t, knn_unscaled.predict(X_test_t))

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train_t)
X_test_scaled = scaler.transform(X_test_t)
knn_scaled = KNeighborsClassifier(n_neighbors=5)
knn_scaled.fit(X_train_scaled, y_train_t)
acc_scaled = accuracy_score(y_test_t, knn_scaled.predict(X_test_scaled))

print("5) KNN accuracy unscaled:", acc_unscaled, "| scaled:", acc_scaled)

# 6) PCA on synthetic data and show explained variance ratio
# Create synthetic high-dimensional data
rng = np.random.RandomState(RANDOM_STATE)
A = rng.normal(size=(500, 10))  # 10 dims
# make first two dimensions informative by adding structure
A[:250, :2] += np.array([2.0, -1.0])
pca = PCA(n_components=10)
pca.fit(A)
print("6) Explained variance ratio (all comps):", np.round(pca.explained_variance_ratio_, 4))
print("   Cumulative explained:", np.round(np.cumsum(pca.explained_variance_ratio_), 4))

# 7) Apply PCA before KNN and compare accuracy (Iris example)
# Keep top 2 components
scaler = StandardScaler()
X_s = scaler.fit_transform(X)
pca2 = PCA(n_components=2)
X_pca = pca2.fit_transform(X_s)

X_pca_train, X_pca_test, y_pca_train, y_pca_test = train_test_split(X_pca, y, test_size=0.2, random_state=RANDOM_STATE, stratify=y)
knn_orig = KNeighborsClassifier(n_neighbors=5)
knn_orig.fit(X_train, y_train)  # from earlier (unscaled Iris)
acc_orig = accuracy_score(y_test, knn_orig.predict(X_test))

knn_pca = KNeighborsClassifier(n_neighbors=5)
knn_pca.fit(X_pca_train, y_pca_train)
acc_pca = accuracy_score(y_pca_test, knn_pca.predict(X_pca_test))
print("7) Iris KNN accuracy original (no PCA, unscaled):", acc_orig)
print("   Iris KNN accuracy after PCA (2 components, scaled data):", acc_pca)

# 8) Hyperparameter tuning on KNN using GridSearchCV (Iris)
param_grid = {'n_neighbors': [1,3,5,7,9], 'weights':['uniform','distance'], 'p':[1,2]}  # p=1 Manhattan, p=2 Euclidean
grid = GridSearchCV(KNeighborsClassifier(), param_grid, cv=5, n_jobs=-1)
grid.fit(X_train, y_train)
print("8) Best GridSearchCV params:", grid.best_params_, "Best CV score:", grid.best_score_)

# 9) Train KNN and count number of misclassified samples (Iris)
knn_final = KNeighborsClassifier(n_neighbors=grid.best_params_['n_neighbors'],
                                 weights=grid.best_params_['weights'],
                                 p=grid.best_params_['p'])
knn_final.fit(X_train, y_train)
y_pred_final = knn_final.predict(X_test)
misclassified = np.sum(y_test != y_pred_final)
print("9) Number of misclassified samples (Iris test):", misclassified)
print("   Confusion matrix:\n", confusion_matrix(y_test, y_pred_final))

# 10) PCA cumulative explained variance visualization (Scree & cumulative)
explained = pca.explained_variance_ratio_
plt.figure(figsize=(8,4))
plt.plot(np.arange(1, len(explained)+1), explained, marker='o', label='Explained ratio')
plt.plot(np.arange(1, len(explained)+1), np.cumsum(explained), marker='x', label='Cumulative')
plt.xlabel('Principal component')
plt.ylabel('Variance ratio')
plt.title('10) Scree plot & cumulative explained variance')
plt.legend()
plt.grid(True)
plt.show()

# ---------------------------
# PRACTICAL (more tasks)
# ---------------------------

# A) KNN with different weights parameter (uniform vs distance) and compare
Xw, yw = load_wine(return_X_y=True)
Xw_train, Xw_test, yw_train, yw_test = train_test_split(Xw, yw, test_size=0.2, random_state=RANDOM_STATE, stratify=yw)
scaler_w = StandardScaler().fit(Xw_train)
Xw_train_s = scaler_w.transform(Xw_train)
Xw_test_s = scaler_w.transform(Xw_test)

for w in ['uniform','distance']:
    knn_w = KNeighborsClassifier(n_neighbors=5, weights=w)
    knn_w.fit(Xw_train_s, yw_train)
    print(f"A) KNN weights={w} accuracy (Wine):", accuracy_score(yw_test, knn_w.predict(Xw_test_s)))

# B) KNN Regressor: effect of different K on performance (MSE)
Xr2, yr2 = make_regression(n_samples=1000, n_features=5, noise=15.0, random_state=RANDOM_STATE)
Xr2_train, Xr2_test, yr2_train, yr2_test = train_test_split(Xr2, yr2, test_size=0.2, random_state=RANDOM_STATE)
for k in [1,3,5,10,20]:
    knr = KNeighborsRegressor(n_neighbors=k)
    knr.fit(Xr2_train, yr2_train)
    print(f"B) KNNReg k={k} MSE:", mean_squared_error(yr2_test, knr.predict(Xr2_test)))

# C) Implement KNN Imputation for missing values
# Create data with missing values
df = pd.DataFrame({
    'A': np.concatenate([np.random.normal(loc=0, scale=1, size=95), np.array([np.nan]*5)]),
    'B': np.concatenate([np.random.normal(loc=2, scale=1, size=90), np.array([np.nan]*10)]),
    'C': np.concatenate([np.random.normal(loc=-1, scale=2, size=100)])
})
# Introduce some random missingness
df.loc[np.random.choice(df.index, 8, replace=False), 'A'] = np.nan
df.loc[np.random.choice(df.index, 6, replace=False), 'B'] = np.nan

print("C) Missing counts before imputation:\n", df.isna().sum())
imputer = KNNImputer(n_neighbors=5)
df_imputed = pd.DataFrame(imputer.fit_transform(df), columns=df.columns)
print("   Missing counts after imputation:\n", pd.DataFrame(df_imputed).isna().sum())

# D) PCA projection to first two principal components (visualization)
cancer = load_breast_cancer()
Xs = StandardScaler().fit_transform(cancer.data)
pca_vis = PCA(n_components=2)
proj = pca_vis.fit_transform(Xs)
plt.figure(figsize=(7,5))
plt.scatter(proj[:,0], proj[:,1], c=cancer.target, cmap='coolwarm', edgecolor='k', s=40)
plt.title('D) PCA projection (breast cancer) onto first 2 PCs')
plt.xlabel('PC1'); plt.ylabel('PC2')
plt.show()

# E) KNN with KD-Tree vs Ball-Tree (compare using algorithm param)
# Use Wine dataset (scaled)
for algo in ['auto','kd_tree','ball_tree']:
    knn_algo = KNeighborsClassifier(n_neighbors=5, algorithm=algo)
    knn_algo.fit(Xw_train_s, yw_train)
    print(f"E) KNN algorithm={algo} accuracy:", accuracy_score(yw_test, knn_algo.predict(Xw_test_s)))

# F) PCA on high-dimensional data and Scree plot (use synthetic 50D)
X_hd = np.random.randn(300, 50)
X_hd[:, :3] += np.linspace(0, 3, 300).reshape(-1,1)  # add some structure in first 3 dims
pca_hd = PCA().fit(X_hd)
plt.figure(figsize=(8,4))
plt.plot(np.arange(1, 51), pca_hd.explained_variance_ratio_, marker='o')
plt.title('F) Scree plot (50D synthetic)')
plt.xlabel('Component'); plt.ylabel('Explained variance ratio')
plt.show()

# G) KNN evaluation using Precision, Recall and F1 (Wine dataset)
knn_g = KNeighborsClassifier(n_neighbors=5)
knn_g.fit(Xw_train_s, yw_train)
y_pred_w = knn_g.predict(Xw_test_s)
prec, rec, f1, _ = precision_recall_fscore_support(yw_test, y_pred_w, average='macro')
print("G) Wine KNN precision, recall, f1 (macro):", (prec, rec, f1))

# H) Effect of different numbers of PCA components on accuracy (Iris)
for n_comp in [1,2,3,4]:
    pca_n = PCA(n_components=n_comp)
    Xn = pca_n.fit_transform(StandardScaler().fit_transform(iris.data))
    Xn_train, Xn_test, yn_train, yn_test = train_test_split(Xn, iris.target, test_size=0.2, random_state=RANDOM_STATE, stratify=iris.target)
    kn = KNeighborsClassifier(n_neighbors=5)
    kn.fit(Xn_train, yn_train)
    print(f"H) PCA n_components={n_comp} Iris KNN acc:", accuracy_score(yn_test, kn.predict(Xn_test)))

# I) KNN leaf_size effect (leaf_size only used by KD-Tree/Ball-Tree implementations)
for leaf in [10, 30, 100]:
    knn_leaf = KNeighborsClassifier(n_neighbors=5, algorithm='kd_tree', leaf_size=leaf)
    knn_leaf.fit(Xw_train_s, yw_train)
    print("I) leaf_size", leaf, "accuracy:", accuracy_score(yw_test, knn_leaf.predict(Xw_test_s)))

# J) Visualize decision boundary for KNN (2D) — already done earlier for Iris first 2 features.
# (Re-show one example)
clf2 = KNeighborsClassifier(n_neighbors=5)
clf2.fit(X2_train, y2_train)
plot_decision_boundary(clf2, X2_train, y2_train, title="J) KNN decision boundary (Iris, k=5, first 2 features)")

# K) KNN on Wine dataset and classification report
knn_wine = KNeighborsClassifier(n_neighbors=5)
knn_wine.fit(Xw_train_s, yw_train)
y_w_pred = knn_wine.predict(Xw_test_s)
print("K) Wine classification report:\n", classification_report(yw_test, y_w_pred, target_names=load_wine().target_names))

# L) KNN regressor with different distance metrics effect (compare MSE)
for metric in ['euclidean', 'manhattan', 'chebyshev']:
    knr_m = KNeighborsRegressor(n_neighbors=5, metric=metric)
    knr_m.fit(Xr2_train, yr2_train)
    print(f"L) KNR metric={metric} MSE:", mean_squared_error(yr2_test, knr_m.predict(Xr2_test)))

# M) KNN ROC-AUC score (for multiclass, show macro average using One-vs-Rest with probability)
# Use Wine dataset (multiclass). We'll compute macro-averaged AUC using binarized labels
from sklearn.preprocessing import label_binarize
y_true_bin = label_binarize(yw_test, classes=np.unique(yw))
probas = knn_wine.predict_proba(Xw_test_s)
# compute macro-average AUC if possible
try:
    roc_auc = roc_auc_score(y_true_bin, probas, average='macro', multi_class='ovr')
    print("M) Wine KNN ROC-AUC (macro OVR):", roc_auc)
except Exception as e:
    print("M) ROC-AUC not available:", e)

# N) PCA: visualize variance captured by each component (already done, but final explicit print)
pca_small = PCA().fit(StandardScaler().fit_transform(iris.data))
print("N) PCA variance ratios (Iris):", np.round(pca_small.explained_variance_ratio_,3))

# O) Feature selection before KNN (simple univariate selection example)
from sklearn.feature_selection import SelectKBest, f_classif
sel = SelectKBest(f_classif, k=3).fit(Xw_train_s, yw_train)
Xw_train_sel = sel.transform(Xw_train_s)
Xw_test_sel = sel.transform(Xw_test_s)
knn_sel = KNeighborsClassifier(n_neighbors=5)
knn_sel.fit(Xw_train_sel, yw_train)
print("O) Wine KNN with top-3 features accuracy:", accuracy_score(yw_test, knn_sel.predict(Xw_test_sel)))

# P) PCA reconstruction error after reducing dimensions
Xorig = StandardScaler().fit_transform(iris.data)
for comps in [1,2,3]:
    pca_r = PCA(n_components=comps)
    Xproj = pca_r.fit_transform(Xorig)
    Xrecon = pca_r.inverse_transform(Xproj)
    recon_err = np.mean((Xorig - Xrecon)**2)
    print(f"P) PCA reconstruction MSE with {comps} comps:", recon_err)

# Done
print("\nAll tasks executed. If you want, I can:")
print("- Export this as a runnable .ipynb (Jupyter) for download.")
print("- Separate scripts per question labeled Q1, Q2, ... to paste into your submission.")
print("- Run or modify any specific block (e.g., change datasets, show plots bigger).")
```
