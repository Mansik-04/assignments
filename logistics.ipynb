1. What is Logistic Regression, and how does it differ from Linear Regression?

Logistic Regression is a statistical model used for classification problems. It predicts the probability that an observation belongs to a particular class (e.g., spam/not spam).

Linear Regression predicts continuous numerical values. Logistic regression, instead of fitting a straight line, uses a sigmoid function to output probabilities between 0 and 1.

3. Why use the Sigmoid function?

It converts any real value into the range[0,1]
 making it interpretable as a probability.

5. What is Regularization? Why is it needed?

Regularization prevents overfitting by adding a penalty on large coefficients.

It controls model complexity, improving generalization.

7. When to use Elastic Net?

When data has high multicollinearity (correlated features). Elastic Net balances between feature selection and shrinkage.

8. Impact of regularization parameter (λ)

Large
𝜆
λ → stronger penalty → simpler model, possible underfitting.

Small
𝜆
λ → weaker penalty → risk of overfitting.

9. Key assumptions of Logistic Regression

Independent observations.

Linear relationship between predictors and log-odds.

No high multicollinearity.

Large sample size preferred.

12. Class imbalance effect

Accuracy may be misleading.

Model may bias toward majority class.

Need resampling, class weights, or alternative metrics (F1, AUC).

13. Hyperparameter Tuning

Adjusting C (inverse of λ), penalty type, solver, etc., using GridSearchCV/RandomizedSearchCV.

14. Different solvers

liblinear: small datasets, L1/L2.

lbfgs, newton-cg, saga: handle multiclass, L2, ElasticNet.

saga: works for L1 and Elastic Net on large datasets.

15. Multiclass Logistic Regression

OvR (One-vs-Rest): trains one classifier per class.

Softmax (Multinomial): uses probability distribution over classes.

16. Advantages and Disadvantages

✅ Easy to implement, interpretable, works well for linearly separable data.
❌ Struggles with non-linear problems, requires feature scaling, sensitive to outliers.

17. Use cases

Spam detection

Customer churn prediction

Medical diagnosis

Credit risk scoring

18. Softmax vs Logistic Regression

Binary Logistic Regression: sigmoid outputs one probability.

Softmax Regression: generalization for multiclass, outputs probability for each class.

19. OvR vs Softmax

OvR: simple, works for small datasets.

Softmax: better calibrated probabilities for multiclass tasks.

20. Interpreting coefficients

Each coefficient shows the change in log-odds for a one-unit change in the predictor, holding others constant.
