```python
# Step 1: Install pymongo
!pip install pymongo[srv] pandas

# Step 2: Import libraries
import pandas as pd
from pymongo import MongoClient

# Step 3: Connect to MongoDB
# ⚠️ Replace with your MongoDB Atlas URI if needed
client = MongoClient("mongodb://localhost:27017/")
db = client["SuperstoreDB"]
orders_col = db["Orders"]

# Step 4: Load dataset into MongoDB
url = "https://drive.google.com/uc?id=1bJ-X2ONfnE5YbsNe2bCK39IfoBHexYQO"
df = pd.read_csv(url, encoding='latin-1')

# Insert only if collection empty
# if orders_col.count_documents({}) == 0:
#     orders_col.insert_many(df.to_dict(orient="records"))

print("\n✅ Superstore data loaded into MongoDB!")
```

```python
# 1. Retrieve all documents
print("\n1. All documents (first 5):")
for doc in orders_col.find().limit(5):
    print(doc)

# 2. Count total documents
print("\n2. Total documents:", orders_col.count_documents({}))

# 3. Orders from West region
print("\n3. Orders from West region:")
for doc in orders_col.find({"Region": "West"}).limit(5):
    print(doc)

# 4. Orders where Sales > 500
print("\n4. Orders with Sales > 500:")
for doc in orders_col.find({"Sales": {"$gt": 500}}).limit(5):
    print(doc)

# 5. Top 3 highest profit
print("\n5. Top 3 orders with highest Profit:")
for doc in orders_col.find().sort("Profit", -1).limit(3):
    print(doc)

# 6. Update Ship Mode
orders_col.update_many({"Ship Mode": "First Class"}, {"$set": {"Ship Mode": "Premium Class"}})
print("\n6. Updated Ship Mode from 'First Class' to 'Premium Class'")

# 7. Delete orders where Sales < 50
orders_col.delete_many({"Sales": {"$lt": 50}})
print("\n7. Deleted orders with Sales < 50")

# 8. Aggregation: Group by Region
print("\n8. Total Sales per Region:")
pipeline = [{"$group": {"_id": "$Region", "TotalSales": {"$sum": "$Sales"}}}]
for doc in orders_col.aggregate(pipeline):
    print(doc)

# 9. Distinct Ship Modes
print("\n9. Distinct Ship Modes:", orders_col.distinct("Ship Mode"))

# 10. Count orders per Category
print("\n10. Orders per Category:")
pipeline = [{"$group": {"_id": "$Category", "Count": {"$sum": 1}}}]
for doc in orders_col.aggregate(pipeline):
    print(doc)
```
