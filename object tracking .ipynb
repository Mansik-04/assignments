```python
import os
import sys
from detectron2.utils.visualizer import Visualizer
from detectron2.data import MetadataCatalog
from detectron2.config import get_cfg
from detectron2.engine import DefaultPredictor
import cv2
from pathlib import Path

# Define placeholder data directories (replace with your actual paths)
TRAIN_DIR = Path('/content/train_data')
DATA_DIR = Path('/content/output_data')

# Check if TRAIN_DIR exists and contains jpg files
jpg_files = list(TRAIN_DIR.glob('*.jpg'))
if not jpg_files:
    print(f"Error: No .jpg files found in {TRAIN_DIR}. Please add some .jpg images to this directory.")
else:
    cfg = get_cfg()
    cfg.MODEL.DEVICE = "cpu" # Set the device to CPU
    cfg.MODEL.WEIGHTS = os.path.join(cfg.OUTPUT_DIR, 'model_final.pth') if os.path.exists(os.path.join(cfg.OUTPUT_DIR, 'model_final.pth')) else cfg.MODEL.WEIGHTS
    cfg.MODEL.ROI_HEADS.SCORE_THRESH_TEST = 0.5
    predictor = DefaultPredictor(cfg)

    sample_img_path = str(jpg_files[0])
    img = cv2.imread(sample_img_path)
    outputs = predictor(img)

    v = Visualizer(img[:, :, ::-1], MetadataCatalog.get('demo_train'), scale=1.2)
    out = v.draw_instance_predictions(outputs['instances'].to('cpu'))

    # Create DATA_DIR if it doesn't exist
    DATA_DIR.mkdir(parents=True, exist_ok=True)

    # Save visualization
    vis_path = DATA_DIR / 'inference_result.jpg'
    cv2.imwrite(str(vis_path), out.get_image()[:, :, ::-1])
    print('Inference visualization saved to', vis_path)
```
