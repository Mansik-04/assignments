```python
🔹 Q1. What is Simple Linear Regression?

Simple Linear Regression is a statistical method to predict the value of a dependent variable (Y) using one independent variable (X).
Equation:

Y=mX+c

m = slope (effect of X on Y)

c = intercept (Y value when X=0)

🔹 Q2. Key assumptions of Simple Linear Regression

Linear relationship between X and Y.

Residuals (errors) are normally distributed.

Homoscedasticity (equal variance of residuals).

Independence of observations.

🔹 Q3. What does coefficient m represent?

It represents slope → how much Y changes if X increases by 1 unit.

🔹 Q4. What does intercept c represent?

It is the value of Y when X = 0.

🔹 Q5. How do we calculate slope m?

Formula:

  m=∑(Xi​−Xˉ)2∑(Xi​−Xˉ)(Yi​−Yˉ)/​∑(Xi​−Xˉ)2


🔹 Q6. Purpose of Least Squares Method

To minimize the sum of squared errors (distance between actual Y and predicted Y).

🔹 Q7. R² interpretation

R² = proportion of variance in Y explained by X.

Example: R² = 0.8 → 80% of Y is explained by X.

🔹 Q8. What is Multiple Linear Regression?

Regression with more than one independent variable (X1, X2, ... Xn).

Y=b0+b1X1+b2X2+...+bnXn

🔹 Q9. Difference: Simple vs Multiple Regression

Simple → 1 predictor

Multiple → many predictors

🔹 Q10. Key assumptions of Multiple Linear Regression

Linearity

No multicollinearity

Homoscedasticity

Independence

Normal distribution of residuals

🔹 Q11. What is heteroscedasticity?

Unequal variance of residuals across values of X.

It weakens reliability of regression results.

🔹 Q12. How to improve model with multicollinearity?

Remove correlated variables

Use regularization (Ridge/Lasso)

Apply PCA (dimensionality reduction)

🔹 Q13. Techniques to transform categorical variables

One-hot encoding (pd.get_dummies)

Label encoding

🔹 Q14. Role of interaction terms

They show combined effect of two variables (e.g., X1 * X2).

🔹 Q15. Intercept interpretation (Simple vs Multiple)

Simple: value of Y when X=0

Multiple: Y when all X=0 (may not always make sense)

🔹 Q16. Significance of slope

Slope shows strength and direction of relationship. Positive slope → increase, Negative slope → decrease.

🔹 Q17. Intercept context

Provides baseline for predictions.

🔹 Q18. Limitations of R²

Always increases with more predictors

Doesn’t show overfitting → use Adjusted R²

🔹 Q19. Large standard error for a coefficient

Means estimate is not reliable (coefficient is unstable).

🔹 Q20. Identifying heteroscedasticity in residual plots

If residuals spread increases/decreases with X → heteroscedasticity present.

🔹 Q21. High R² but low Adjusted R²

Model has many predictors, but most are not useful → overfitting.

🔹 Q22. Why scale variables?

Because regression coefficients depend on scale. Scaling improves interpretation & convergence.

🔹 Q23. What is Polynomial Regression?

Regression where independent variable is raised to powers.

Y=b0+b1X+b2X2+...+bnXn

🔹 Q24. Difference between polynomial & linear regression

Linear regression → straight line

Polynomial regression → curve

🔹 Q25. When to use Polynomial Regression?

When data shows non-linear relationship.

🔹 Q26. General equation of Polynomial Regression

Y=b0+b1X+b2X2+...+bnXn

🔹 Q27. Can polynomial regression be applied to multiple variables?

Yes → Polynomial terms of each variable + interactions.

🔹 Q28. Limitations of Polynomial Regression

Overfitting

High complexity with large degree

🔹 Q29. Methods to evaluate polynomial degree

Cross-validation

Adjusted R²

AIC/BIC

🔹 Q30. Why visualization is important?

To check if polynomial curve actually fits data properly.

🔹 Q31. Polynomial Regression in Python
# Example: Polynomial Regression
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures

# Sample data
X = np.array([1, 2, 3, 4, 5]).reshape(-1,1)
y = np.array([2, 5, 10, 17, 26])  # Quadratic relation

# Transform to polynomial features
poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)

# Fit regression model
model = LinearRegression()
model.fit(X_poly, y)

# Predictions
y_pred = model.predict(X_poly)

# Plot
plt.scatter(X, y, color="blue")
plt.plot(X, y_pred, color="red")
plt.show()
```
